```python
df['Price category'] = pd.cut(df['Price'], bins=[-1, 200, 400, 3000],labels=['Low', 'Medium', 'High'])
```

```python
df['Rating score'] = df['Avg ratings'] * df['Total ratings']
```

```python
df['Delivery time'] = df['Delivery time'].astype(float)
```

```python
df.head(5)
```

```python
print(df.isnull().sum())
```

```python
num_cols = df.select_dtypes(include='number').columns
df[num_cols] = df[num_cols].fillna(df[num_cols].mean())
```

```python
city_stats = df.groupby('City').agg(restaurants=('ID', 'count'),avg_price=('Price', 'mean'),avg_rating=('Avg ratings', 'mean'),avg_delivery=('Delivery time', 'mean'))
print(city_stats.sort_values('restaurants', ascending=False))
```
